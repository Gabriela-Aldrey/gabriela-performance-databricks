
## Overview

This notebook will show you how to create and query a table or DataFrame that you uploaded to DBFS. [DBFS](https://docs.databricks.com/user-guide/dbfs-databricks-file-system.html) is a Databricks File System that allows you to store data for querying inside of Databricks. This notebook assumes that you have a file already inside of DBFS that you would like to read from.

This notebook is written in **Python** so the default cell type is Python. However, you can use different languages by using the `%LANGUAGE` syntax. Python, Scala, SQL, and R are all supported.

```
# File location and type
file_location = "/FileStore/tables/student_por.csv"
file_type = "csv"

# CSV options
infer_schema = "false"
first_row_is_header = "false"
delimiter = ","

# The applied options are for CSV files. For other file types, these will be ignored.
df = spark.read.format(file_type) \
  .option("inferSchema", infer_schema) \
  .option("header", first_row_is_header) \
  .option("sep", delimiter) \
  .load(file_location)

display(df)
```

```
# Create a view or table

temp_table_name = "student_por_csv"

df.createOrReplaceTempView(temp_table_name)
```

```
# Leitura dos dois arquivos CSV (português)
df_por = spark.read.option("header", True).option("inferSchema", True).csv("/FileStore/tables/student_por.csv")

df_total = df_por
df_total.display()
```

```
from pyspark.sql.functions import monotonically_increasing_id

# Criando tabela de dimensão aluno
df_dim_aluno = df_total.select(
    "sex", "age", "famsize", "Pstatus", "Medu", "Fedu", "studytime", 
    "activities", "schoolsup", "famsup", "paid", "higher", 
    "internet", "romantic", "famrel", "freetime", "goout", 
    "Dalc", "Walc", "health"
).dropDuplicates()

# Adicionando uma coluna de ID
df_dim_aluno = df_dim_aluno.withColumn("id_dim_aluno", monotonically_increasing_id())
df_dim_aluno.display()
```

```
# Junção com df_total para trazer o id_dim_aluno
df_fato = df_total.join(df_dim_aluno, on=[
    "sex", "age", "famsize", "Pstatus", "Medu", "Fedu", "studytime", 
    "activities", "schoolsup", "famsup", "paid", "higher", 
    "internet", "romantic", "famrel", "freetime", "goout", 
    "Dalc", "Walc", "health"
], how="left")

df_fato = df_fato.select(
    "id_dim_aluno", "G1", "G2", "G3", "absences"
).withColumnRenamed("G1", "nota_g1") \
 .withColumnRenamed("G2", "nota_g2") \
 .withColumnRenamed("G3", "nota_g3") \
 .withColumnRenamed("absences", "faltas")

df_fato.display()
```

```
# Salvando dimensão aluno
df_dim_aluno.write.format("delta").mode("overwrite").save("/delta/dim_aluno")
spark.sql("DROP TABLE IF EXISTS dim_aluno")
spark.sql("CREATE TABLE dim_aluno USING DELTA LOCATION '/delta/dim_aluno'")

# Salvando tabela fato
df_fato.write.format("delta").mode("overwrite").save("/delta/fato_desempenho")
spark.sql("DROP TABLE IF EXISTS fato_desempenho")
spark.sql("CREATE TABLE fato_desempenho USING DELTA LOCATION '/delta/fato_desempenho'")
```

1. Pessoas do sexo feminino possuem médias superiores ao masculinho ? - Sim!

```
%sql
SELECT sex, ROUND(AVG(nota_g3), 2) AS media_final
FROM fato_desempenho f
JOIN dim_aluno d ON f.id_dim_aluno = d.id_dim_aluno
GROUP BY sex;
```

2. Quanto maior o tempo estudando, melhor será a média final? SIm!

```
%sql
SELECT studytime, ROUND(AVG(nota_g3), 2) AS media_final
FROM fato_desempenho f
JOIN dim_aluno d ON f.id_dim_aluno = d.id_dim_aluno
GROUP BY studytime
ORDER BY studytime;
```

```
%sql
SELECT Medu, ROUND(AVG(nota_g3), 2) AS media_mae,
       Fedu, ROUND(AVG(nota_g3), 2) AS media_pai
FROM fato_desempenho f
JOIN dim_aluno d ON f.id_dim_aluno = d.id_dim_aluno
GROUP BY Medu, Fedu
ORDER BY Medu DESC, Fedu DESC;
```

Alunos que trabalham tem média inferior aos que não trabalham ?

```
%sql
SELECT workday, ROUND(AVG(nota_g3), 2) AS media_final
FROM (
    SELECT *,
           CASE WHEN paid = 'yes' THEN 'Sim' ELSE 'Não' END AS workday
    FROM dim_aluno
) a
JOIN fato_desempenho f ON a.id_dim_aluno = f.id_dim_aluno
GROUP BY workday;
```

```
%sql
SELECT schoolsup, ROUND(AVG(nota_g3), 2) AS media_final
FROM fato_desempenho f
JOIN dim_aluno d ON f.id_dim_aluno = d.id_dim_aluno
GROUP BY schoolsup;
```

6. Atividades Extracurriculares ajudam ?

```
%sql
SELECT activities, ROUND(AVG(nota_g3), 2) AS media_final
FROM fato_desempenho f
JOIN dim_aluno d ON f.id_dim_aluno = d.id_dim_aluno
GROUP BY activities;
```

7. Faltas prejudicam o desempenho? - Sim! Quanto mais faltas, pior o desempenho


```
%sql
SELECT CASE 
           WHEN faltas <= 5 THEN '0-5'
           WHEN faltas <= 10 THEN '6-10'
           WHEN faltas <= 20 THEN '11-20'
           ELSE '21+' 
       END AS grupo_faltas,
       ROUND(AVG(nota_g3), 2) AS media_final
FROM fato_desempenho
GROUP BY grupo_faltas
ORDER BY grupo_faltas;
```

8. Consumo de álcool afeta o rendimento? - Sim! Quanto maior o consumo de álcool, menor média final

```
%sql
SELECT Dalc, Walc, ROUND(AVG(nota_g3), 2) AS media_final
FROM fato_desempenho f
JOIN dim_aluno d ON f.id_dim_aluno = d.id_dim_aluno
GROUP BY Dalc, Walc
ORDER BY Dalc, Walc;
```
